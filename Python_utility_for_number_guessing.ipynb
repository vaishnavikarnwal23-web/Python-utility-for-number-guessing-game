```python
import random

# Function to display the welcome message
def welcome_message():
  print("=" * 40)
  print(" NUMBER GUESSING SYSTEM")
  print("=" * 40)
  print("TRY TO GUESS THE NUMBER SELECTED BY THE COMPUTER.")
  print()


# Function to select difficulty
def choose_difficulty():
  print("Choose Difficulty level:")
  print("1. Easy (1-50)")
  print("2. Medium (1-100)")
  print("3.Hard (1-200)")

  while True:

    choice = input("Enter your choice(1/2/3):")

    if choice == "1":
      return 50, "Easy"
    elif choice == "2":
      return 100, "Medium"
    elif choice == "3":
      return 200, "Hard"
    else:
      print("Invalid choice.Please choose 1,2 or 3.")


# Function to get a valid guess
def get_guess(maximum):

  while True:

    try:
      guess = int(input("Enter your guess:"))

      if guess < 1 or guess > maximum:
        print("Please enter a number between 1 and", maximum)
      else:
        return guess

    except ValueError:
      print("Please enter a valid number.")


# Function to calculate score
def calculate_score(attempts):

  score = 100 - ((attempts - 1)* 10)

  if score < 0:
    score = 0

  return score


# Function to play the game
def play_game():

  maximum, difficulty = choose_difficulty()

  # Computer selects a random number
  secret_number = random.randint(1,maximum)


  attempts = 0

  print()
  print("I have selected a number between 1 and", maximum)
  print("Start guessing:")
  print()

  while True:

    guess = get_guess(maximum)

    attempts = attempts + 1

    # Compare the guess with secret number
    if guess < secret_number:

      print("Too Low! Try a higher number: ")

    elif guess > secret_number:

      print("Too High! Try a lower number: ")

    else:

      print()
      print("Congratulations!")
      print("You guessed the correct number ")

      score = calculate_score(attempts)


      print("Number of attempts: ", attempts)
      print("Your score:", score)

      return score


# Main program
welcome_message()

best_score = 0

while True:

  score = play_game()

  #check best score
  if score > best_score:
    best_score = score
    print("New Best Score!")
  else:
    print("Best Score:", best_score)

  print()
  play_again = input("Do you want to play again? (yes/no):")

  if play_again.lower() == "no":

    print()
    print("="* 40)
    print("Thankyou for playing!")
    print("Your best score was:", best_score)
    print("="* 40)

    break

  elif play_again.lower() == "yes":

    print()
    print("Starting a new game...")
    print()


  else:
    print()
    print("Invalid choice.")
    print("Game ended")
    break


```
